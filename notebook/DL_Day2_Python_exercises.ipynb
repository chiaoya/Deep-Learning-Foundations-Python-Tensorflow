### A) Python-Grundlagen

**A1.** `Ergebnis: 3   Rest: 2`

*Anmerkung:* `//` rundet immer ab (Ganzzahldivision), `%` gibt den Rest zurück — beide zusammen "zerlegen" eine Zahl in ganze Teile und Rest.

```python
print((6 + (3 * 5)) // 4)  # (6 + 15) // 4 = 21 // 4 = 5
```

*Anmerkung:* Erst die Klammer, dann `*`, dann `//`. Punkt-vor-Strich gilt auch in Python.

```python
stadt = "Hamburg"
einwohner = 1900000
ist_hauptstadt = False
print("In " + stadt + " leben " + str(einwohner) + " Menschen. Hauptstadt? " + str(ist_hauptstadt))
```

*Anmerkung:* Zahlen und Booleans müssen beim Verbinden mit `+` immer zuerst mit `str()` in Text umgewandelt werden — sonst gibt es einen `TypeError`.

```python
zahl = 17
if zahl % 2 == 0:
    print(zahl, "ist gerade")
else:
    print(zahl, "ist ungerade")
```

*Anmerkung:* `% 2 == 0` ist der Standardtrick für "gerade" — kommt in der Praxis ständig vor (z. B. jede zweite Zeile einfärben).

```python
for i in range(20, -1, -5):
    print(i)
```

*Anmerkung:* Bei negativer Schrittweite muss der Stopp-Wert kleiner als der Start sein, sonst läuft die Schleife gar nicht.

```python
x = 8
while x > 0:
    print(x)
    x -= 1
```

*Anmerkung:* Ohne `x -= 1` läuft die Schleife für immer (Endlosschleife) — die häufigste Fehlerquelle bei `while`.

### B) Pandas

```python
konzerte["Jahre_her"] = 2026 - konzerte["Jahr"]
konzerte.head()
```

```python
konzerte.sort_values(by="Zuschauer", ascending=False)
```

```python
konzerte_voll = pd.merge(konzerte, staedte, on="Stadt_ID", how="inner")
konzerte_voll.head()
```

**B5.** `"inner"` behält nur Zeilen, deren Schlüssel in beiden Tabellen vorkommt. `"outer"` behält alle Zeilen aus beiden Tabellen und füllt fehlende Werte mit `NaN` auf.

*Anmerkung:* `"inner"` ist der Standard und meistens das, was man will. `"outer"` hilft, wenn man auch sehen will, was NICHT zusammenpasst.

### C) Datenvisualisierung

**C1.** Histogramm · Balkendiagramm · Streudiagramm (Scatterplot) · Boxplot

```python
zuschauer_pro_band = konzerte.groupby("Band")["Zuschauer"].mean()

plt.figure(figsize=(7,4))
plt.bar(zuschauer_pro_band.index, zuschauer_pro_band.values)
plt.title("Ø Zuschauer pro Band")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(data=konzerte_voll, x="Stadt", y="Ticketpreis")
plt.title("Ticketpreis nach Stadt")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.scatterplot(data=konzerte, x="Ticketpreis", y="Zufriedenheit")
plt.title("Ticketpreis vs. Zufriedenheit")
plt.tight_layout()
plt.show()
```

*Anmerkung:* Falls ein positiver Zusammenhang sichtbar ist (höherer Preis → höhere Zufriedenheit): Korrelation ist nicht gleich Kausalität! Ein hoher Preis "verursacht" nicht automatisch Zufriedenheit — es könnte z. B. sein, dass teurere Konzerte auch bekanntere Bands haben.
